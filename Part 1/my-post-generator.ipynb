### Import Libraries & API Keys

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import display, Markdown

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
if OPENAI_API_KEY is None:
    raise Exception("API Key is missing")
```

### Step 2: Searching the Web

```python
client = OpenAI()
completion = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role": "user", "content": "What is today's date? Tell me the top 3 stories in the past week related to music, according to reliable news media. Each story should be from a seperate source and date. Include dates."}
    ],
    max_tokens=500
)
```

```python
search_results = completion.choices[0].message.content
display(Markdown(search_results))
```
