```python
from litellm import completion
```

```python
messages = [{"role": "user", "content": "What are the most sought after tech skills in 2026?"}]

response = completion(
    model="openai/gpt-5-nano",
    messages=messages
)

reply = response.choices[0].message.content
display(Markdown(reply))
```
