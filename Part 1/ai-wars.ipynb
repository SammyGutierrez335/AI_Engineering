### Import Libraries & API Keys

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import display, Markdown

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
if OPENAI_API_KEY is None:
    raise Exception("API Key is missing")
```

### Step 2: AI Battle

```python
client = OpenAI()

gpt_good_model = 'gpt-4.1-mini'
gpt_evil_model = 'gpt-4.1-nano'

gpt_good_system_prompt = """You are helpful and ethical assistant. You goal is to provide accurate, safe, and respectful information. 
You always proritize a users well being. Keep your responses concise and to the point. """

gpt_evil_system_prompt = """You are mischeivous and unhelpful assistant. You goal is to confuse users and provide misinformation. 
You always proritize CHAOS. Keep your responses concise and to the point. """

good_messages = [
    {"role": "system", "content": gpt_good_system_prompt}
]
evil_messages = [
    {"role": "system", "content": gpt_evil_system_prompt}
]

new_message_good = "Hello"

display(Markdown(f"Good-\n {new_message_good}"))
for i in range(3):
    evil_messages.append({"role": "user", "content": new_message_good})
    good_messages.append({"role": "assistant", "content": new_message_good})

    #evil model response
    evil_completion = client.chat.completions.create(
        model=gpt_evil_model,
        messages=evil_messages
)
    evil_response = evil_completion.choices[0].message.content
    display(Markdown(f"Evil-\n {evil_response}"))
    evil_messages.append({"role": "assistant", "content": evil_response})
    good_messages.append({"role": "user", "content": evil_response})

    good_completion = client.chat.completions.create(
           model=gpt_good_model,
           messages=good_messages
    )
    new_message_good = good_completion.choices[0].message.content
    display(Markdown(f"Good-\n {new_message_good}"))
```
