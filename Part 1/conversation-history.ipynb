### Step 1: Import Libraries and load the environment variables

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv
OPENAI_API_KEY = os.getenv('OPENAI_API_KEY')

if OPENAI_API_KEY is None:
    raise Exception("API key is missing")
else:
    print("Hello World!")

client = OpenAI()
```

### Step 2: Call the API

```python
responses = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        { "role": "user", "content": "You are a helpful assistant." },
        { "role": "user", "content": "Next week I will be traveling to Spain." }
    ]
)

reply = responses.choices[0].message.content


print(reply)
```

### In Theory...(Doing it manually to demonstrate Psuedo Memory)

```python
messages=[
        { "role": "user", "content": "You are a helpful assistant." },
        { "role": "user", "content": "Sure I can use tips." },
        { "role": "assistant", "content": "That sounds exciting! Do you need any help with travel tips, places to visit, or anything related to your trip to Spain?"},
        { "role": "user", "content": "Yes, I would like to know the best restaraunts"}
]

responses = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages
)

reply = responses.choices[0].message.content

print(reply)
```
