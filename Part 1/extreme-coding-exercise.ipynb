### import API
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv
OPENAI_API_KEY = os.getenv('OPENAI_API_KEY')

if OPENAI_API_KEY is None:
    raise Exception("API key is missing")

client = OpenAI()
completions = client.chat.completions.create({
    model = "gpt-4.1-mini",
    messages = [{"role": "system", "content": "you are a helpful assistant"}, {"role": "user", "content": "Is this the correct way to write an api call" }]
})

response = completions.messages[0].message.content
