### Import Libraries and load the environment variables

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv
OPENAI_API_KEY = os.getenv('OPENAI_API_KEY')

if OPENAI_API_KEY is None:
    raise Exception("API key is missing")
else:
    print("Hello World!")

client = OpenAI()
```

### 1. Responses API: basic call


```python
response = client.responses.create(
    model="gpt-4.1-mini",
    instructions="You are a helpful assistant",
    input="Next week I will be traveling to Spain."
    # before
    # messages=[
    #     { "role": "system", "content": "You are a helpful assistant." },
    #     { "role": "user", "content": "Next week I will be traveling to Spain." }
    # ]
)
reply = response.output_text
display(Markdown(reply))
```

### 2. Responses API: conversation history

```python
response = client.responses.create(
    model="gpt-4.1-mini",
    instructions="You are a helpful assistant",
    input="Next week I will be traveling to Spain."
)

previous_id = response.id

reply = response.output_text
display(Markdown(reply))
```

```python
response = client.responses.create(
    model="gpt-4.1-mini",
    previous_response_id=previous_id,
    instructions="You are a helpful assistant",
    input="Where am I traveling to next week?.",
    # store = true is default to make it statefull (stores on server for 30 days)
    store=false # removes storing of converstation.
)
previous_id = response.id

reply = response.output_text
display(Markdown(reply))
```

```python
response = client.responses.create(
    model="gpt-4.1-mini",
    previous_response_id=previous_id,
    instructions="You are a helpful assistant",
    input="What is the capital?."
)
previous_id = response.id
reply = response.output_text
display(Markdown(reply))
```

### 3. Responses API: Built-in Tools

```python
tone="dramatic and engaging"

instructions = f"""
You are a helpful assistant who creates amazing LinkedIn post about AI on
text that you are provided with. Make the post engaging and interesting.
You will be supplied some instructions by the user, based on that input write a post.
Also, make concise and add a thought provoking question at the end to encourage comments.
Use teh following tone: {tone}
"""
prompt = "Write about the 3 top stories in AI this past week, according to reliable news media. Include dates"

resp = client.responses.create(
    model = "gpt-6-luna",
    instructions=instructions,
    input=prompt,
    tools=[{"type": "web_search"}],
)

display(Markdown(resp.output_text))
```
