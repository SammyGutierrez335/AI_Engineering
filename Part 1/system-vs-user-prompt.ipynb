### Step 1: Import Libraries and load the environment variables

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv
OPENAI_API_KEY = os.getenv('OPENAI_API_KEY')

if OPENAI_API_KEY is None:
    raise Exception("API key is missing")
else:
    print("Hello World!")

client = OpenAI()
```

### Step 2: Call the API

```python
responses = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[{
        "role": "user", "content": "Hi there my, name is Sammy"
    }]
)

reply = responses.choices[0].message.content

print(responses)
```

### Step 3a - Calling the API with a System Prompt

```python
responses = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role": "system", "content": "You are a 5 star restaurant chef assisting other 5 star chef's how to cook your award winning pasta"}, #system prompt
        {"role": "user", "content": "How do I cook pasta?"}
    ]
)

reply = responses.choices[0].message.content

# print(reply)
display(Markdown(reply))
```

### 3b: Changing the system prompt changes the LLM behavior

```python
responses = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role": "system", "content": "You are a mischievious assistant who always gives incorrect and misleading advice"}, #system prompt
        {"role": "user", "content": "How do I cook pasta?"}
        ]
)

reply = responses.choices[0].message.content

# print(reply)
display(Markdown(reply))
```

### Commercial Use-case: IKEA Customer Service Agent

```python
inquiry = "I bought the desk 365 days ago but lost my receipt. May I return it?"
ikeaReturnPolicy = "At IKEA, it’s OK to change your mind. Sometimes what seemed perfect for your home, office or garden, just isn’t. Returning a product to us is convenient and hassle-free, no matter how or where it’s bought. You have 365 days to return your purchase for a full refund. We will refund you by the same method as your original payment. If the returns criteria are not fully met, we may refuse the return or offer an exchange or gift card."
```

```python
responses = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        { "role": "system", "content": f"You are a customer service agent for Ikea. You are helpful and provide polite customer service but strictly follow the return policies of the store. Our return policy is: {ikeaReturnPolicy}"},
        { "role": "user", "content": inquiry }
    ]
)

reply = responses.choices[0].message.content
display(Markdown(reply))
```
