```python
import random

from openai import OpenAI

client = OpenAI()

def roll_dice():
    return random.randint(1,6)
```

```python
target = 6
max_iterations = 20
```

```python
history = []
iteration = 0

while iteration < max_iterations:
    instructions = f"""
        You are playing a dice game. You need to roll a {target} on a dice.
        You've made the follow rolls so far. 
        {history if history else "No rolls yet."}
    
        Based on what you have rolled so far, decide your next action: 

        Option A: If you have achieved teh target, respond with exactly: "DONE: <message describing how you achieved the target>"
        Option B: If you have not achieved the target, respond with exactly: ROLL THE DICE
    """
    response = client.chat.completions.create(
        model="gpt-4.1-mini",
        messages=[
            {"role": "user", "content": instructions}
        ]
    )

    iteration += 1
    result = response.choices[0].message.content
    if "DONE" in result:
        print(result)
        break

    roll = roll_dice()
    print("rolling...")
    print(f"Rolled: {roll} \n")


    if "ROLL THE DICE" in result:
        history += f"Roll {iteration: {roll}}"

print(f"Game ended after {iteration} iteration")
    #...process the result, to see if the agent has decided whether to roll again or not.
    #...stop the loop if the agent has decided to stop OR if max_iterations is reached.
```
